```python
import numpy as np
def array_factory(mode , shape ,value=None):
    if mode=='zeros':
        np.zeros(shape)
    elif mode=='ones':
        np.ones(shape)
    elif mode=='full':
        np.full(shape,value)
    elif mode=='identity':
        np.eye(shape)
    else:
        raise ValueError('ivalid mode')
      
```

```python
import numpy as np
def secure_reshape_and_stack(data1,data2,new_shape):
    try:
        arr1=np.array(data1)
        arr2=np.array(data2)
        reshaped_arr1=arr1.reshape(new_shape)
        if reshaped_arr1.shape[1]!=arr2.shape[1]:
            raise ValueError('arrays columns number must be the same')
        combined_dataset=np.vstack((reshaped_arr1,arr2))
        return combined_dataset
    except ValueError as e:
        raise ValueError(f'comapny grade error:{e}')
        
```
